# Overview

Notebook mirror of the Streamlit Overview tab. It reads generated CSV/PNG outputs from the pipeline.

```python
from pathlib import Path
import pandas as pd
from IPython.display import Image, display

ROOT = Path('..').resolve()
CSV_DIR = ROOT / 'results' / 'csv'
FIG_DIR = ROOT / 'results' / 'figures'

def read_csv(name):
    path = CSV_DIR / name
    return pd.read_csv(path) if path.exists() else pd.DataFrame()

def show_fig(name):
    path = FIG_DIR / name
    if path.exists():
        display(Image(filename=str(path)))
```
