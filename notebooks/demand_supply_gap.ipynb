```python
from pathlib import Path
import pandas as pd
from IPython.display import Image, display

ROOT = Path('..').resolve()
CSV_DIR = ROOT / 'results' / 'csv'
FIG_DIR = ROOT / 'results' / 'figures'
DOMAIN = 'IT'

def slug(domain):
    return domain.lower().replace(' ', '_')

domain_slug = slug(DOMAIN)
```

```python
relevance = pd.read_csv(CSV_DIR / f'{domain_slug}_skill_relevance.csv')
gap = pd.read_csv(CSV_DIR / f'{domain_slug}_skill_gap.csv')
relevance.head(20)
```
