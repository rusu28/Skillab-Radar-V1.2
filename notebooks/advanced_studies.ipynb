# Advanced Studies

Notebook mirror of the Streamlit Advanced Studies tab, including Skill Shock Index, Career Bridge, and Training ROI Priority.

```python
from pathlib import Path
import pandas as pd
from IPython.display import Image, display

ROOT = Path('..').resolve()
CSV_DIR = ROOT / 'results' / 'csv'
FIG_DIR = ROOT / 'results' / 'figures'

studies = [
    ('Domain Market Attractiveness', 'advanced_domain_market_attractiveness.csv', 'advanced_domain_market_attractiveness.png'),
    ('Domain Choice Radar', 'advanced_domain_market_attractiveness.csv', 'advanced_domain_choice_radar.png'),
    ('Skill Shock Index', 'advanced_skill_shock_index.csv', 'advanced_skill_shock_index.png'),
    ('Career Bridge / Reskilling Pathways', 'advanced_career_bridges.csv', 'advanced_career_bridges.png'),
    ('Training ROI Priority', 'advanced_training_roi_priority.csv', 'advanced_training_roi_priority.png'),
    ('Future Risk Quadrant', 'advanced_future_risk_quadrant.csv', 'advanced_future_risk_quadrant.png'),
    ('Training Alignment', 'advanced_training_alignment.csv', 'advanced_training_alignment.png'),
    ('ESCO Capability Shortages', 'advanced_capability_family_summary.csv', 'advanced_capability_family_gaps.png'),
    ('Domain Signal Scoreboard', 'advanced_domain_signal_scoreboard.csv', 'advanced_domain_signal_scoreboard.png'),
]
```

```python
for title, csv_name, fig_name in studies:
    print('\n' + '=' * 80)
    print(title)
    csv_path = CSV_DIR / csv_name
    if csv_path.exists():
        display(pd.read_csv(csv_path).head(15))
    fig_path = FIG_DIR / fig_name
    if fig_path.exists():
        display(Image(filename=str(fig_path)))
```
