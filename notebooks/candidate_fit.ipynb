# Candidate Fit and Reskilling Pathways

Notebook mirror of the Streamlit Candidate Fit tab.

```python
from pathlib import Path
import pandas as pd
from IPython.display import Image, display

ROOT = Path('..').resolve()
CSV_DIR = ROOT / 'results' / 'csv'
FIG_DIR = ROOT / 'results' / 'figures'
DOMAIN = 'IT'
domain_slug = DOMAIN.lower().replace(' ', '_')
```

Top candidate score is the weighted overlap between a profile's skills and the selected domain's priority skills. Missing priority skills form the upskilling explanation.

```python
candidates = pd.read_csv(CSV_DIR / f'{domain_slug}_candidate_matches.csv')
candidates.head(20)
```

```python
readiness = pd.read_csv(CSV_DIR / 'advanced_candidate_readiness.csv')
readiness[readiness['domain'] == DOMAIN].head(20)
```

## Career Bridge / Reskilling Pathways

These rows show profiles from other domains that are closest to the selected target domain. `missing_reskilling_skill_labels` is the suggested learning path.

```python
bridges = pd.read_csv(CSV_DIR / 'advanced_career_bridges.csv')
bridges[bridges['target_domain'] == DOMAIN].head(20)
```
